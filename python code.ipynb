```python
# Step 1: Import Required Libraries

import pandas as pd
import numpy as np
```

```python
# Step 2: Load Dataset

df = pd.read_excel('Ecommerce_Unclean_Project.xlsx')
```

```python
# Step 3: Explore the Dataset
# View First 5 Records

df.head()
```

```python
df.info()
```

```python
#Statistical Summery

df.describe()
```

```python
# Dataset Shape

df.shape
```

```python
# Step 4: Check Missing Values

df.isnull().sum()
```

```python
# Step 5: Check Duplicate Records

df.duplicated().sum()
```

```python
# Remove Duplicate Records

df.drop_duplicates(inplace=True)
```

```python
df.duplicated().sum()
```

```python
# Step 6: Replace Invalid Values
# Replace common invalid values with NoN

df.replace(['N/A','NULL',''],np.nan,inplace=True)
```

```python
# Step 7: Remove Extra Spaces

df = df.apply(lambda x: x.str.strip() if x.dtype =='object' else x)
```

```python
# Step 8: Standardize Text Format
# Convert text into Proper Case.

df['Customer_Name'] = df['Customer_Name'].str.title()

df['City'] = df['City'].str.title()

df['State'] = df['State'].str.title()
```

```python
# Step 9: Validate Email Addresses
# Keep only valid email records.

df = df[df['Email'].str.contains('@',na=False)]
```

```python
# Step 10: Convert Date Columns

df['Order_Date']=pd.to_datetime(df['Order_Date'],errors='coerce')
df['Delivery_Date']=pd.to_datetime(df['Delivery_Date'],errors='coerce')
```

```python
# Step 11: Convert Numeric Columns

cols = ['Qty','Unit_Price','Discount']

for c in cols:
    df[c] = pd.to_numeric(df[c],errors='coerce')
```

```python
# Step 12: Remove Invlid Quantity
# Remove records having negative or zero quantity.

df = df[df['Qty']>0]
```

```python
# Step 13: Fill Missing Values

df['Discount'] = df['Discount'].fillna(0)

df['Phone'] = df['Phone'].fillna('Unknown')

df['Delivery_Date'] = df['Delivery_Date'].fillna(df['Order_Date'])
```

```python
# Step 14: Feature Engineering
# Total Sales

df['Sales']=df['Qty']*df['Unit_Price']
```

```python
df
```

```python
# Recovering missing discount values..

original_df = pd.read_excel('Ecommerce_Unclean_Project.xlsx')
```

```python
original_df['Discount'].value_counts(dropna=False)
```

```python
df['Discount'] = original_df['Discount'].fillna(0)
```

```python
# Net Amount

df['Discount'] = pd.to_numeric(df['Discount'], errors='coerce').fillna(0)
df['Net_Amount'] = df['Sales'] - (df['Sales'] * df['Discount'] / 100)
```

```python
df
```

```python
# Profit (Example: 20%)
df['Profit']=df['Net_Amount']*0.20
```

```python
df
```

```python
# Order Month

df['Month']=df['Order_Date'].dt.month_name()
```

```python
df
```

```python
# Order Year

df['Year']=df['Order_Date'].dt.year
```

```python
df
```

```python
# Weekday

df['Weekday']=df['Order_Date'].dt.day_name()
```

```python
df
```

```python
# Step 16: Export Clean Data

df.to_csv('Clean_Ecommerce.csv',index=False)
```
